Replace all odd numbers in the array

```python
def replace_odd(replace, array):
    # function to replace odd nmbers with a constant
    for i in range(0,len(array)):
        if array[i] % 2 != 0:
            array[i] = replace
    return array

replace_odd(1,oned)
```

Find the position of:

elements in x where its value is more than its corresponding element in y and elements in x where its value is equal to its corresponding in y

```python
def compare_array(x,y):
    # finds the index where x = y and x< y
    n = min(len(x),len(y))
    same = []
    more = []
    for i in range(0,n):
        if x[i] < y[i]:
            same.append(i)
        if x[i] == y[i]:
            more.append(i)
    return same,more
compare_array(oned,np.array([2,3,0,1,6,1])) 
```
